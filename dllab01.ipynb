```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

from mpl_toolkits.mplot3d import Axes3D
```

```python
# Generate a random 4x4 array from exponential distribution
exp_array = np.random.exponential(scale=1.0, size=(4,4))

print("Random 4x4 Exponential Distribution Array:")
print(exp_array)
```

```python
# Generate random data
exp_data = np.random.exponential(scale=1.0, size=100000)
uniform_data = np.random.rand(100000)
normal_data = np.random.randn(100000)

# Plot histograms
plt.figure(figsize=(10,6))

plt.hist(exp_data,
         bins=100,
         density=True,
         histtype='step',
         color='red',
         label='Exponential')

plt.hist(uniform_data,
         bins=100,
         density=True,
         histtype='step',
         color='blue',
         label='Uniform')

plt.hist(normal_data,
         bins=100,
         density=True,
         histtype='step',
         color='green',
         label='Normal')

plt.xlim(-4,8)
plt.ylim(0,1.2)

plt.title("Comparison of Random Distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Create X and Y values
x = np.linspace(-5,5,100)
y = np.linspace(-5,5,100)

X, Y = np.meshgrid(x,y)

# Calculate Z
Z = X**2 + Y**2

# Plot
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_title("3D Plot of Z = X² + Y²")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```

```python
!wget -q https://elitedatascience.com/wp-content/uploads/2022/07/Pokemon.csv

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Pokemon.csv", index_col=0, encoding="latin")

df.head()
```

```python
features = ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]

new_df = df[features]

new_df.head()
```

```python
pearson_corr = new_df.corr(method="pearson")

plt.figure(figsize=(8,6))
sns.heatmap(pearson_corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Pearson Correlation Heatmap")
plt.show()
```

```python
spearman_corr = new_df.corr(method="spearman")

plt.figure(figsize=(8,6))
sns.heatmap(spearman_corr, annot=True, cmap="YlGnBu", fmt=".2f")

plt.title("Spearman Correlation Heatmap")
plt.show()
```
